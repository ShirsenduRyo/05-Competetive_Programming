```python
def productExceptSelf(nums):
    n = len(nums)
    answer = [1] * n

    # Step 1: Prefix products
    prefix = 1
    print("Prefix products:")
    for i in range(n):
        answer[i] = prefix
        prefix *= nums[i]
        print(f"Index {i}: prefix={prefix}, answer={answer}")

    # Step 2: Suffix products (on the fly)
    suffix = 1
    print("\nSuffix products:")
    for i in range(n-1, -1, -1):
        answer[i] *= suffix
        suffix *= nums[i]
        print(f"Index {i}: suffix={suffix}, answer={answer}")

    return answer
```

```python
print(productExceptSelf([1, 2, 3, 4]))  # Output: [24, 12, 8, 6]
```
